### Como usar o `type()` em Python
A função `type()` nos diz a qual classe ou tipo de dado pertence um valor. Veja abaixo alguns exemplos comuns:

```python
nome = 'Luan'          # Texto (String) -> str
idade = 15            # Número inteiro -> int
peso = 58.8           # Número decimal -> float

# Para ver o tipo de cada um, usamos o print(type(variavel))
print(type(nome))
print(type(idade))
print(type(peso))
```

```python
a = input('digite algo:')
print(f'O tipo primitivo desse valor é {type(a)}')
print(f'Só tem espaços? {a.isspace()}')
print(f'É um número? {a.isnumeric()}')
print(f'É alfabético? {a.isalpha()}')
print(f'e alfanumérico? {a.isalnum()}')
print(f'eata em maiusculas? {a.isupper()}')
print(f'esta em minusculas? {a.islower()}')
print(f'esta capitalizda? {a.istitle()}')
```
